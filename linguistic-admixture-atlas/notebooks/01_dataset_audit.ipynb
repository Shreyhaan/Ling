# Dataset audit: South/Central Asian pilot

This notebook audits `data_processed/languages_master.csv`, prints the Grambank/PHOIBLE overlap counts, and draws the first longitude/latitude scatterplot.

```python
import pandas as pd

df = pd.read_csv("../data_processed/languages_master.csv")

print(df.shape)
display(df.head())
```

```python
for col in ["in_grambank", "in_phoible"]:
    df[col] = df[col].astype(str).str.lower().eq("true")

print("In Grambank:", df["in_grambank"].sum())
print("In PHOIBLE:", df["in_phoible"].sum())
print("In both:", (df["in_grambank"] & df["in_phoible"]).sum())

df[df["in_grambank"] & df["in_phoible"]].head(30)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))
plt.scatter(df["longitude"], df["latitude"], s=8)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Candidate South/Central Asian language sample")
plt.show()
```
